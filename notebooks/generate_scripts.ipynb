```python
folder_name = 'run_scripts_cifar'
run_file = 'run_cifar.py'

model = 'mom_sgd'
iid_s = [True, False]

aggs_s = ['avg', 'cm', 'rfa']
bucketing_s = [0, 2, 2]

attack_s = ['ALIE', 'BF', 'LF', 'NA', 'IPM', 'SHB']

lr_s = [0.01, 0.001]

seed_s = [123]

batch_size = 128
test_batch_size = 1024
epochs = 30

workers = 20
byzantines = 5

compression = 'none'

partial_participation = 0.20
clip_update_s = [True, False]

clip_mult_s = [0.1, 1., 10.]

# set directory one level up
folder_name = f'../{folder_name}'
```

```python
import os 
from itertools import product

os.makedirs(folder_name, exist_ok=True)

exp_name = run_file.split('.')[0].split('_')[1]

for iid, attack, (agg, bucketing) in product(iid_s, attack_s, zip(aggs_s, bucketing_s)):
    
    file_name = f'{folder_name}/{exp_name}_{agg}_{attack.lower()}_{"iid" if iid else "niid"}.sh'
    with open(file_name, 'w') as f:
        f.write(f'#!/bin/bash\n\n')
    for lr, seed, clip_update in product(lr_s, seed_s, clip_update_s):
        for clip_mult in clip_mult_s:

            clip_update_str = f' --clip-update --clip-mult {clip_mult}' if clip_update else ''
            iid_str = ' --shuffle' if iid else ''
            with open(file_name, 'a+') as f:
                # continue writing
                f.write(f'python {run_file} --model {model} --agg {agg} --bucketing {bucketing}'
                        f' --attack {attack} --lr {lr} --seed {seed} -b {batch_size}'
                        f' --test-batch-size {test_batch_size} -e {epochs} -n {workers} -f {byzantines}'
                        f' --use-cuda --eval-every 1 --compression {compression} --partial-participation'
                        f' --partial-participation-ratio {partial_participation}'
                        f'{clip_update_str}{iid_str}\n')
            
            if not clip_update:
                break
    f.close()
    os.system(f'chmod +x {file_name}')
    print(f'File {file_name} created')
   
```
